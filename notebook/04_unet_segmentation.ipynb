```python
# ============================================================
# 04_unet_segmentation_256.py
# Textile Defect Detection - U-Net Segmentation
# 256 x 256 training version
# ============================================================

from pathlib import Path
import shutil
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt


# ============================================================
# 1. PATHS
# ============================================================

PROJECT_DIR = Path(r"D:\proto\TextileDefectPrototype")

DATASET_DIR = PROJECT_DIR / "dataset"
DEFECT_DIR = DATASET_DIR / "defect"

MODEL_DIR = PROJECT_DIR / "models"
NOTEBOOK_MODEL_DIR = PROJECT_DIR / "notebook" / "models"

MODEL_DIR.mkdir(parents=True, exist_ok=True)
NOTEBOOK_MODEL_DIR.mkdir(parents=True, exist_ok=True)

BEST_MODEL_PATH = NOTEBOOK_MODEL_DIR / "unet_256_best.keras"
FINAL_MODEL_PATH = MODEL_DIR / "unet_segmentation.keras"


# ============================================================
# 2. SETTINGS
# ============================================================

IMG_SIZE = (256, 256)

BATCH_SIZE = 8
EPOCHS = 8

RANDOM_STATE = 42

AUTOTUNE = tf.data.AUTOTUNE


# ============================================================
# 3. REPRODUCIBILITY
# ============================================================

tf.keras.utils.set_random_seed(RANDOM_STATE)

np.random.seed(RANDOM_STATE)


# ============================================================
# 4. FIND PAIRED IMAGES AND MASKS
# ============================================================
#
# Only JPG/JPEG files are used as input images.
# PNG files are used only as ground-truth masks.
#
# Example:
#
# T1-D2-028.jpg  -> input image
# T1-D2-028.png  -> ground-truth mask
#
# ============================================================

image_paths = []
mask_paths = []

for image_path in sorted(DEFECT_DIR.glob("*.jpg")):

    mask_path = DEFECT_DIR / f"{image_path.stem}.png"

    if mask_path.exists():

        image_paths.append(
            str(image_path)
        )

        mask_paths.append(
            str(mask_path)
        )


for image_path in sorted(DEFECT_DIR.glob("*.jpeg")):

    mask_path = DEFECT_DIR / f"{image_path.stem}.png"

    if mask_path.exists():

        image_paths.append(
            str(image_path)
        )

        mask_paths.append(
            str(mask_path)
        )


print("==========================================")
print("U-NET DATASET")
print("==========================================")

print(
    "Total paired images:",
    len(image_paths)
)

print("==========================================")


if len(image_paths) == 0:

    raise RuntimeError(
        "No paired JPG/JPEG + PNG files were found."
    )


# ============================================================
# 5. TRAIN / VALIDATION SPLIT
# ============================================================

indices = np.arange(
    len(image_paths)
)

rng = np.random.default_rng(
    RANDOM_STATE
)

rng.shuffle(
    indices
)

split_index = int(
    len(indices) * 0.85
)

train_indices = indices[
    :split_index
]

val_indices = indices[
    split_index:
]


train_image_paths = [
    image_paths[i]
    for i in train_indices
]

train_mask_paths = [
    mask_paths[i]
    for i in train_indices
]


val_image_paths = [
    image_paths[i]
    for i in val_indices
]

val_mask_paths = [
    mask_paths[i]
    for i in val_indices
]


print(
    "Training images:",
    len(train_image_paths)
)

print(
    "Validation images:",
    len(val_image_paths)
)


# ============================================================
# 6. IMAGE + MASK LOADER
# ============================================================

def load_image_mask(
    image_path,
    mask_path
):

    # --------------------------------------------------------
    # IMAGE
    # --------------------------------------------------------

    image = tf.io.read_file(
        image_path
    )

    image = tf.image.decode_jpeg(
        image,
        channels=3
    )

    image = tf.image.resize(
        image,
        IMG_SIZE,
        method="bilinear"
    )

    image = tf.cast(
        image,
        tf.float32
    ) / 255.0


    # --------------------------------------------------------
    # MASK
    # --------------------------------------------------------

    mask = tf.io.read_file(
        mask_path
    )

    mask = tf.image.decode_png(
        mask,
        channels=1
    )

    mask = tf.image.resize(
        mask,
        IMG_SIZE,
        method="nearest"
    )

    mask = tf.cast(
        mask,
        tf.float32
    ) / 255.0

    mask = tf.where(
        mask > 0.5,
        1.0,
        0.0
    )

    return image, mask


# ============================================================
# 7. CREATE TF.DATASETS
# ============================================================

train_dataset = tf.data.Dataset.from_tensor_slices(
    (
        train_image_paths,
        train_mask_paths
    )
)

train_dataset = train_dataset.map(
    load_image_mask,
    num_parallel_calls=AUTOTUNE
)

train_dataset = train_dataset.shuffle(
    buffer_size=len(train_image_paths),
    seed=RANDOM_STATE,
    reshuffle_each_iteration=True
)

train_dataset = train_dataset.batch(
    BATCH_SIZE
)

train_dataset = train_dataset.prefetch(
    AUTOTUNE
)


val_dataset = tf.data.Dataset.from_tensor_slices(
    (
        val_image_paths,
        val_mask_paths
    )
)

val_dataset = val_dataset.map(
    load_image_mask,
    num_parallel_calls=AUTOTUNE
)

val_dataset = val_dataset.batch(
    BATCH_SIZE
)

val_dataset = val_dataset.prefetch(
    AUTOTUNE
)


print("==========================================")
print("DATASETS READY")
print("==========================================")

print(
    "Train batches:",
    tf.data.experimental.cardinality(
        train_dataset
    ).numpy()
)

print(
    "Validation batches:",
    tf.data.experimental.cardinality(
        val_dataset
    ).numpy()
)


# ============================================================
# 8. DICE METRIC
# ============================================================

def dice_metric(
    y_true,
    y_pred
):

    y_true = tf.cast(
        y_true,
        tf.float32
    )

    y_pred = tf.cast(
        y_pred,
        tf.float32
    )

    intersection = tf.reduce_sum(
        y_true * y_pred,
        axis=[1, 2, 3]
    )

    denominator = (
        tf.reduce_sum(
            y_true,
            axis=[1, 2, 3]
        )
        +
        tf.reduce_sum(
            y_pred,
            axis=[1, 2, 3]
        )
    )

    dice = (
        2.0 * intersection + 1e-7
    ) / (
        denominator + 1e-7
    )

    return tf.reduce_mean(
        dice
    )


# ============================================================
# 9. IoU METRIC
# ============================================================

def iou_metric(
    y_true,
    y_pred
):

    y_true = tf.cast(
        y_true,
        tf.float32
    )

    y_pred = tf.cast(
        y_pred,
        tf.float32
    )

    intersection = tf.reduce_sum(
        y_true * y_pred,
        axis=[1, 2, 3]
    )

    union = (
        tf.reduce_sum(
            y_true,
            axis=[1, 2, 3]
        )
        +
        tf.reduce_sum(
            y_pred,
            axis=[1, 2, 3]
        )
        -
        intersection
    )

    iou = (
        intersection + 1e-7
    ) / (
        union + 1e-7
    )

    return tf.reduce_mean(
        iou
    )


# ============================================================
# 10. DICE LOSS
# ============================================================

def dice_loss(
    y_true,
    y_pred
):

    return 1.0 - dice_metric(
        y_true,
        y_pred
    )


# ============================================================
# 11. COMBINED BCE + DICE LOSS
# ============================================================

def combined_loss(
    y_true,
    y_pred
):

    bce = tf.keras.losses.binary_crossentropy(
        y_true,
        y_pred
    )

    bce = tf.reduce_mean(
        bce
    )

    return (
        bce
        +
        dice_loss(
            y_true,
            y_pred
        )
    )


# ============================================================
# 12. U-NET BLOCKS
# ============================================================

def conv_block(
    x,
    filters
):

    x = tf.keras.layers.Conv2D(
        filters,
        3,
        padding="same"
    )(x)

    x = tf.keras.layers.BatchNormalization()(
        x
    )

    x = tf.keras.layers.Activation(
        "relu"
    )(x)

    x = tf.keras.layers.Conv2D(
        filters,
        3,
        padding="same"
    )(x)

    x = tf.keras.layers.BatchNormalization()(
        x
    )

    x = tf.keras.layers.Activation(
        "relu"
    )(x)

    return x


# ============================================================
# 13. BUILD U-NET
# ============================================================

def build_unet():

    inputs = tf.keras.Input(
        shape=(
            IMG_SIZE[0],
            IMG_SIZE[1],
            3
        ),
        name="input_image"
    )


    # --------------------------------------------------------
    # ENCODER
    # --------------------------------------------------------

    c1 = conv_block(
        inputs,
        32
    )

    p1 = tf.keras.layers.MaxPooling2D(
        2
    )(c1)


    c2 = conv_block(
        p1,
        64
    )

    p2 = tf.keras.layers.MaxPooling2D(
        2
    )(c2)


    c3 = conv_block(
        p2,
        128
    )

    p3 = tf.keras.layers.MaxPooling2D(
        2
    )(c3)


    c4 = conv_block(
        p3,
        256
    )

    p4 = tf.keras.layers.MaxPooling2D(
        2
    )(c4)


    # --------------------------------------------------------
    # BOTTLENECK
    # --------------------------------------------------------

    bottleneck = conv_block(
        p4,
        512
    )


    # --------------------------------------------------------
    # DECODER
    # --------------------------------------------------------

    u4 = tf.keras.layers.Conv2DTranspose(
        256,
        2,
        strides=2,
        padding="same"
    )(bottleneck)

    u4 = tf.keras.layers.Concatenate()(
        [u4, c4]
    )

    c5 = conv_block(
        u4,
        256
    )


    u3 = tf.keras.layers.Conv2DTranspose(
        128,
        2,
        strides=2,
        padding="same"
    )(c5)

    u3 = tf.keras.layers.Concatenate()(
        [u3, c3]
    )

    c6 = conv_block(
        u3,
        128
    )


    u2 = tf.keras.layers.Conv2DTranspose(
        64,
        2,
        strides=2,
        padding="same"
    )(c6)

    u2 = tf.keras.layers.Concatenate()(
        [u2, c2]
    )

    c7 = conv_block(
        u2,
        64
    )


    u1 = tf.keras.layers.Conv2DTranspose(
        32,
        2,
        strides=2,
        padding="same"
    )(c7)

    u1 = tf.keras.layers.Concatenate()(
        [u1, c1]
    )

    c8 = conv_block(
        u1,
        32
    )


    # --------------------------------------------------------
    # OUTPUT
    # --------------------------------------------------------

    outputs = tf.keras.layers.Conv2D(
        1,
        1,
        activation="sigmoid",
        name="defect_probability"
    )(c8)


    return tf.keras.Model(
        inputs,
        outputs,
        name="Textile_U_Net_256"
    )


# ============================================================
# 14. CREATE MODEL
# ============================================================

unet_model = build_unet()

unet_model.summary()


# ============================================================
# 15. COMPILE MODEL
# ============================================================

unet_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-4
    ),
    loss=combined_loss,
    metrics=[
        dice_metric,
        iou_metric
    ]
)


# ============================================================
# 16. CALLBACKS
# ============================================================

checkpoint = tf.keras.callbacks.ModelCheckpoint(
    filepath=str(
        BEST_MODEL_PATH
    ),
    monitor="val_iou_metric",
    mode="max",
    save_best_only=True,
    verbose=1
)


early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_iou_metric",
    mode="max",
    patience=4,
    restore_best_weights=True,
    verbose=1
)


reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(
    monitor="val_iou_metric",
    mode="max",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)


# ============================================================
# 17. TRAIN
# ============================================================

print("==========================================")
print("STARTING U-NET TRAINING")
print("==========================================")

history = unet_model.fit(
    train_dataset,
    validation_data=val_dataset,
    epochs=EPOCHS,
    callbacks=[
        checkpoint,
        early_stopping,
        reduce_lr
    ]
)


# ============================================================
# 18. LOAD BEST MODEL
# ============================================================

print("==========================================")
print("LOADING BEST MODEL")
print("==========================================")

unet_model = tf.keras.models.load_model(
    BEST_MODEL_PATH,
    custom_objects={
        "combined_loss": combined_loss,
        "dice_metric": dice_metric,
        "iou_metric": iou_metric
    }
)


# ============================================================
# 19. SAVE FINAL MODEL FOR FLASK
# ============================================================

unet_model.save(
    FINAL_MODEL_PATH
)

print(
    "Final U-Net saved to:",
    FINAL_MODEL_PATH
)


# ============================================================
# 20. TRAINING CURVES
# ============================================================

plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Loss"
)

plt.title(
    "U-Net Training Loss"
)

plt.legend()

plt.grid(
    True
)

plt.show()


plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["dice_metric"],
    label="Training Dice"
)

plt.plot(
    history.history["val_dice_metric"],
    label="Validation Dice"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "Dice"
)

plt.title(
    "U-Net Dice"
)

plt.legend()

plt.grid(
    True
)

plt.show()


plt.figure(
    figsize=(8, 5)
)

plt.plot(
    history.history["iou_metric"],
    label="Training IoU"
)

plt.plot(
    history.history["val_iou_metric"],
    label="Validation IoU"
)

plt.xlabel(
    "Epoch"
)

plt.ylabel(
    "IoU"
)

plt.title(
    "U-Net IoU"
)

plt.legend()

plt.grid(
    True
)

plt.show()


# ============================================================
# 21. NUMPY IoU
# ============================================================

def numpy_iou(
    ground_truth,
    prediction
):

    ground_truth = (
        ground_truth > 0
    )

    prediction = (
        prediction > 0
    )

    intersection = np.logical_and(
        ground_truth,
        prediction
    ).sum()

    union = np.logical_or(
        ground_truth,
        prediction
    ).sum()

    if union == 0:

        if intersection == 0:
            return 1.0

        return 0.0

    return (
        intersection / union
    )


# ============================================================
# 22. NUMPY DICE
# ============================================================

def numpy_dice(
    ground_truth,
    prediction
):

    ground_truth = (
        ground_truth > 0
    )

    prediction = (
        prediction > 0
    )

    intersection = np.logical_and(
        ground_truth,
        prediction
    ).sum()

    denominator = (
        ground_truth.sum()
        +
        prediction.sum()
    )

    if denominator == 0:

        if intersection == 0:
            return 1.0

        return 0.0

    return (
        2.0 * intersection
    ) / denominator


# ============================================================
# 23. GET VALIDATION ARRAYS
# ============================================================

print("==========================================")
print("GENERATING VALIDATION PREDICTIONS")
print("==========================================")


validation_images = []
validation_masks = []


for image_batch, mask_batch in val_dataset:

    validation_images.append(
        image_batch.numpy()
    )

    validation_masks.append(
        mask_batch.numpy()
    )


validation_images = np.concatenate(
    validation_images,
    axis=0
)

validation_masks = np.concatenate(
    validation_masks,
    axis=0
)


print(
    "Validation image array:",
    validation_images.shape
)

print(
    "Validation mask array:",
    validation_masks.shape
)


# ============================================================
# 24. PREDICT VALIDATION SET
# ============================================================

validation_predictions = unet_model.predict(
    validation_images,
    batch_size=BATCH_SIZE,
    verbose=1
)


validation_predictions = (
    validation_predictions[:, :, :, 0]
)

validation_masks = (
    validation_masks[:, :, :, 0]
)


# ============================================================
# 25. THRESHOLD SWEEP
# ============================================================

thresholds = [
    0.05,
    0.10,
    0.15,
    0.20,
    0.25,
    0.30,
    0.40,
    0.50
]


threshold_results = []


print("==========================================")
print("U-NET THRESHOLD SWEEP")
print("==========================================")


for threshold in thresholds:

    iou_scores = []
    dice_scores = []

    for index in range(
        len(validation_predictions)
    ):

        predicted_binary = (
            validation_predictions[index]
            >= threshold
        )

        ground_truth_binary = (
            validation_masks[index]
            > 0.5
        )

        iou_scores.append(
            numpy_iou(
                ground_truth_binary,
                predicted_binary
            )
        )

        dice_scores.append(
            numpy_dice(
                ground_truth_binary,
                predicted_binary
            )
        )


    mean_iou = np.mean(
        iou_scores
    )

    mean_dice = np.mean(
        dice_scores
    )


    threshold_results.append(
        (
            threshold,
            mean_iou,
            mean_dice
        )
    )


    print(
        f"Threshold {threshold:.2f} | "
        f"IoU {mean_iou * 100:.2f}% | "
        f"Dice {mean_dice * 100:.2f}%"
    )


# ============================================================
# 26. SELECT BEST THRESHOLD BY IoU
# ============================================================

best_threshold_result = max(
    threshold_results,
    key=lambda item: item[1]
)

BEST_THRESHOLD = (
    best_threshold_result[0]
)

BEST_IOU = (
    best_threshold_result[1]
)

BEST_DICE_AT_IOU_THRESHOLD = (
    best_threshold_result[2]
)


print("==========================================")
print("BEST U-NET THRESHOLD")
print("==========================================")

print(
    "Best threshold:",
    BEST_THRESHOLD
)

print(
    "Best IoU:",
    round(
        BEST_IOU * 100,
        2
    ),
    "%"
)

print(
    "Dice at best IoU threshold:",
    round(
        BEST_DICE_AT_IOU_THRESHOLD * 100,
        2
    ),
    "%"
)


# ============================================================
# 27. BEST THRESHOLD BY DICE
# ============================================================

best_dice_result = max(
    threshold_results,
    key=lambda item: item[2]
)

print(
    "Best Dice threshold:",
    best_dice_result[0]
)

print(
    "Best Dice:",
    round(
        best_dice_result[2] * 100,
        2
    ),
    "%"
)

print(
    "IoU at best Dice threshold:",
    round(
        best_dice_result[1] * 100,
        2
    ),
    "%"
)


# ============================================================
# 28. VISUALIZE VALIDATION EXAMPLES
# ============================================================

DISPLAY_THRESHOLD = BEST_THRESHOLD

num_examples = min(
    5,
    len(validation_images)
)


for index in range(
    num_examples
):

    image = validation_images[
        index
    ]

    ground_truth = validation_masks[
        index
    ]

    prediction = validation_predictions[
        index
    ]

    predicted_binary = (
        prediction
        >= DISPLAY_THRESHOLD
    )


    plt.figure(
        figsize=(15, 4)
    )


    plt.subplot(
        1,
        3,
        1
    )

    plt.imshow(
        image
    )

    plt.title(
        "Original"
    )

    plt.axis(
        "off"
    )


    plt.subplot(
        1,
        3,
        2
    )

    plt.imshow(
        ground_truth,
        cmap="gray"
    )

    plt.title(
        "Ground Truth"
    )

    plt.axis(
        "off"
    )


    plt.subplot(
        1,
        3,
        3
    )

    plt.imshow(
        predicted_binary,
        cmap="gray"
    )

    plt.title(
        f"Prediction @ {DISPLAY_THRESHOLD:.2f}"
    )

    plt.axis(
        "off"
    )


    plt.tight_layout()

    plt.show()


# ============================================================
# 29. TEST ONE ORIGINAL DATASET IMAGE
# ============================================================

TEST_IMAGE_PATH = (
    DEFECT_DIR /
    "T1-D2-001.jpg"
)


if TEST_IMAGE_PATH.exists():

    test_image, test_mask = (
        load_image_mask(
            str(TEST_IMAGE_PATH),
            str(
                DEFECT_DIR /
                "T1-D2-001.png"
            )
        )
    )

    test_image_array = (
        test_image.numpy()
    )

    test_mask_array = (
        test_mask.numpy()[:, :, 0]
    )


    test_prediction = unet_model.predict(
        np.expand_dims(
            test_image_array,
            axis=0
        ),
        verbose=0
    )[0, :, :, 0]


    test_binary = (
        test_prediction
        >= BEST_THRESHOLD
    )


    test_iou = numpy_iou(
        test_mask_array,
        test_binary
    )

    test_dice = numpy_dice(
        test_mask_array,
        test_binary
    )

    test_area = (
        np.mean(
            test_binary
        )
        * 100
    )


    print("==========================================")
    print("T1-D2-001 TEST")
    print("==========================================")

    print(
        "Prediction minimum:",
        float(
            test_prediction.min()
        )
    )

    print(
        "Prediction maximum:",
        float(
            test_prediction.max()
        )
    )

    print(
        "Prediction mean:",
        float(
            test_prediction.mean()
        )
    )

    print(
        "Threshold:",
        BEST_THRESHOLD
    )

    print(
        "IoU:",
        round(
            test_iou * 100,
            2
        ),
        "%"
    )

    print(
        "Dice:",
        round(
            test_dice * 100,
            2
        ),
        "%"
    )

    print(
        "Defect area:",
        round(
            test_area,
            2
        ),
        "%"
    )

    print("==========================================")


# ============================================================
# 30. SAVE TRAINING INFORMATION
# ============================================================

threshold_file = (
    MODEL_DIR /
    "unet_256_threshold.txt"
)

threshold_file.write_text(
    (
        f"Best threshold: {BEST_THRESHOLD}\n"
        f"Best validation IoU: {BEST_IOU * 100:.4f}%\n"
        f"Dice at best IoU threshold: "
        f"{BEST_DICE_AT_IOU_THRESHOLD * 100:.4f}%\n"
        f"Best Dice threshold: {best_dice_result[0]}\n"
        f"Best validation Dice: {best_dice_result[2] * 100:.4f}%\n"
        f"IoU at best Dice threshold: {best_dice_result[1] * 100:.4f}%\n"
        f"Image size: {IMG_SIZE}\n"
        f"Training images: {len(train_image_paths)}\n"
        f"Validation images: {len(val_image_paths)}\n"
    ),
    encoding="utf-8"
)


print("==========================================")
print("TRAINING COMPLETE")
print("==========================================")

print(
    "Best model:",
    BEST_MODEL_PATH
)

print(
    "Flask model:",
    FINAL_MODEL_PATH
)

print(
    "Threshold file:",
    threshold_file
)

print(
    "Best inference threshold:",
    BEST_THRESHOLD
)

print("==========================================")
```
